```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

pd.set_option('display.max_columns', None)

print("All libraries imported successfully!")
```

```python
df = pd.read_csv(
    r"C:\Users\mukun\Downloads\TrafficDataset\Metro_Interstate_Traffic_Volume.csv"
)

print("Dataset loaded successfully!")
```

```python
df['holiday'] = df['holiday'].fillna('None')

print("Missing values handled successfully!")
```

```python
df.isnull().sum()
```

```python
df['date_time'] = pd.to_datetime(df['date_time'])

print("date_time converted successfully!")
```

```python
df['date_time'].dtype
```

```python
df['hour'] = df['date_time'].dt.hour
df['day'] = df['date_time'].dt.day
df['month'] = df['date_time'].dt.month
df['year'] = df['date_time'].dt.year
df['day_of_week'] = df['date_time'].dt.dayofweek
df['day_name'] = df['date_time'].dt.day_name()

df['is_weekend'] = df['day_of_week'] >= 5

print("Time features created successfully!")
```

```python
df.head()
```

```python
df['temp_celsius'] = df['temp'] - 273.15

print("Temperature converted to Celsius!")
```

```python
df[['temp', 'temp_celsius']].head()
```

```python
df[['traffic_volume', 'temp_celsius', 'rain_1h', 'snow_1h', 'clouds_all']].describe()
```

```python
hourly_traffic = df.groupby('hour')['traffic_volume'].mean()

plt.figure(figsize=(10, 5))

plt.plot(
    hourly_traffic.index,
    hourly_traffic.values,
    marker='o'
)

plt.title('Average Traffic Volume by Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Average Traffic Volume')

plt.xticks(range(24))
plt.grid(True, alpha=0.3)

plt.show()
```

```python
day_traffic = df.groupby('day_name')['traffic_volume'].mean()

# Arrange days in the correct order
day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

day_traffic = day_traffic.reindex(day_order)

plt.figure(figsize=(10, 5))

plt.bar(day_traffic.index, day_traffic.values)

plt.title('Average Traffic Volume by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Average Traffic Volume')

plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
weekend_traffic = df.groupby('is_weekend')['traffic_volume'].mean()

weekend_traffic.index = ['Weekday', 'Weekend']

plt.figure(figsize=(7, 5))

plt.bar(weekend_traffic.index, weekend_traffic.values)

plt.title('Average Traffic Volume: Weekday vs Weekend')
plt.xlabel('Day Type')
plt.ylabel('Average Traffic Volume')

plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
weather_traffic = df.groupby('weather_main')['traffic_volume'].mean()
weather_traffic = weather_traffic.sort_values(ascending=False)

plt.figure(figsize=(10, 5))

plt.bar(weather_traffic.index, weather_traffic.values)

plt.title('Average Traffic Volume by Weather Condition')
plt.xlabel('Weather Condition')
plt.ylabel('Average Traffic Volume')

plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    df['rain_1h'],
    df['traffic_volume'],
    alpha=0.3
)

plt.title('Rainfall vs Traffic Volume')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Traffic Volume')

plt.grid(True, alpha=0.3)

plt.show()
```

```python
# Replace the invalid temperature value
df.loc[df['temp_celsius'] == -273.15, 'temp_celsius'] = np.nan

# Fill the missing temperature with the median
df['temp_celsius'] = df['temp_celsius'].fillna(
    df['temp_celsius'].median()
)

print("Invalid temperature values handled successfully!")
```
